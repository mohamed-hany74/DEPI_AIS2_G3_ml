```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
```

```python
df =pd.DataFrame(data=data['data'],columns=data.feature_names)
```

```python
df.head(5)
```

```python
df['target'] = data['target']
df.head(5)
```

```python
df['target'].value_counts()
```

```python
x = df.iloc[:,:-1].values 
y = df['target'].values
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
def sigmoid (x) :
    return 1/(1+np.exp(-x))
```

```python
def  update_weighs(weights , x , z ,y , lr) :
    m = len(y)
    dw = (1/m) *  np.dot(x.T,(z-y))
    weights = weights - lr*dw
    return weights
```

```python
def  update_baiss(baiss , x , z ,y , lr) :
    m = len(y)
    db = (1/m) *  np.sum(z-y)
    baiss = baiss - lr*db
    return baiss
```

```python
def perceptron(x,weights , baiss) : 
    z = np.dot(x, weights ) + baiss
    y_pred = sigmoid(z)
    y_p = [1 if y>0.5 else 0 for y in y_pred]
    return y_p
```

```python
def accuracy(y_pred , y_actual) :
    return np.sum(y_pred == y_actual)/len(y_actual)
```

```python
lr =0.001 
weights = np.zeros(x_train.shape[1])
bais = 0
for i in range(20000):
 pred = perceptron(x_train, weights, bais)
 weights = update_weighs(weights, x_train, pred, y_train, lr)
 bais = update_baiss(bais, x_train, pred, y_train, lr)

```

```python
y_pred = perceptron(x_train, weights, bais)
print("Accuracy", accuracy(y_pred, y_train))
```

```python
y_pred = perceptron(x_test, weights, bais)
```

```python
print("Accuracy", accuracy(y_pred, y_test))
```
